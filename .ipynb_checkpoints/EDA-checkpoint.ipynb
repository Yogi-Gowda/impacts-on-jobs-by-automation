```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import warnings
warnings.filterwarnings('ignore')

plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')

OUTPUT_DIR = '../visualizations'
os.makedirs(OUTPUT_DIR, exist_ok=True)

X_train = pd.read_csv('data/X_train.csv')

y_train = pd.read_csv('data/y_train.csv')

df = pd.concat([X_train, y_train], axis=1)

print(f"X_train  : {X_train.shape[0]} rows, {X_train.shape[1]} columns")
print(f"y_train  : {y_train.shape[0]} rows, {y_train.shape[1]} column(s)")
print(f"Combined : {df.shape[0]} rows, {df.shape[1]} columns")
print(f"\nColumns: {list(df.columns)}")


print("\n BASIC DATASET INFORMATION")

# Shape
print(f"\nRows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

# Data types
print("\nData Types:")
print(df.dtypes)

# Missing values
print("\nMissing Values:")
missing = df.isnull().sum()
if missing.sum() == 0:
    print("  No missing values found!")
else:
    print(missing[missing > 0])

# Duplicate rows
duplicates = df.duplicated().sum()
print(f"\nDuplicate Rows: {duplicates}")

# Summary statistics
print("\nStatistical Summary:")
print(df.describe())

print("\n" + "=" * 65)
print("UNIVARIATE ANALYSIS")
print("=" * 65)

# Define which columns are numerical and which are categorical
numerical_cols = [
    'Median Salary (USD)',
    'Experience Required (Years)',
    'Job Openings (2024)',
    'Projected Openings (2030)',
    'Remote Work Ratio (%)',
    'Automation Risk (%)',
    'Gender Diversity (%)'
]

categorical_cols = [
    'Industry',
    'Job Status',
    'AI Impact Level',
    'Required Education',
    'Location'
]

# Plot 1: Target variable distribution 
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram: shows how many jobs fall in each risk range
axes[0].hist(df['Automation Risk (%)'], bins=30, color='skyblue', edgecolor='black')
axes[0].axvline(df['Automation Risk (%)'].mean(), color='red', linestyle='--',
                label=f"Mean: {df['Automation Risk (%)'].mean():.1f}%")
axes[0].set_xlabel('Automation Risk (%)')
axes[0].set_ylabel('Number of Jobs')
axes[0].set_title('Distribution of Automation Risk')
axes[0].legend()

# Box plot: shows median, spread, and outliers
axes[1].boxplot(df['Automation Risk (%)'])
axes[1].set_ylabel('Automation Risk (%)')
axes[1].set_title('Box Plot of Automation Risk')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/01_automation_risk_distribution.png', dpi=150, bbox_inches='tight')
plt.show()

# Plot 2: Histograms for all numerical columns 
fig, axes = plt.subplots(3, 3, figsize=(18, 14))
axes = axes.ravel()  # flatten from 3x3 grid to a simple list

for i, col in enumerate(numerical_cols):
    axes[i].hist(df[col], bins=30, color='lightcoral', edgecolor='black', alpha=0.8)
    axes[i].axvline(df[col].mean(), color='blue', linestyle='--',
                    label=f'Mean: {df[col].mean():.1f}')
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')
    axes[i].legend(fontsize=8)

# Hide unused subplot panels
for i in range(len(numerical_cols), len(axes)):
    fig.delaxes(axes[i])

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/02_numerical_distributions.png', dpi=150, bbox_inches='tight')
plt.show()

# Plot 3: Bar charts for all categorical columns 
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.ravel()

for i, col in enumerate(categorical_cols):
    counts = df[col].value_counts()  # count rows per category

    axes[i].bar(range(len(counts)), counts.values,
                color=plt.cm.Set3(np.linspace(0, 1, len(counts))))
    axes[i].set_xticks(range(len(counts)))
    axes[i].set_xticklabels(counts.index, rotation=45, ha='right', fontsize=9)
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_ylabel('Count')

    # Add count labels on top of each bar
    for j, v in enumerate(counts.values):
        axes[i].text(j, v + 50, str(v), ha='center', va='bottom', fontsize=8)

fig.delaxes(axes[5])  # hide the unused last panel
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/03_categorical_distributions.png', dpi=150, bbox_inches='tight')
plt.show()

print("Univariate analysis done!")



print("\n BIVARIATE ANALYSIS")


#  Plot 4: Risk by Industry 
# Calculate average risk per industry, sorted highest to lowest
industry_risk = df.groupby('Industry')['Automation Risk (%)'].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(14, 6))
ax.bar(range(len(industry_risk)), industry_risk.values,
       color=plt.cm.RdYlGn_r(industry_risk.values / 100))  # red=high, green=low
ax.set_xticks(range(len(industry_risk)))
ax.set_xticklabels(industry_risk.index, rotation=45, ha='right')
ax.set_ylabel('Average Automation Risk (%)')
ax.set_title('Average Automation Risk by Industry')
ax.grid(axis='y', alpha=0.3)

for i, v in enumerate(industry_risk.values):
    ax.text(i, v + 0.5, f'{v:.1f}%', ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/04_risk_by_industry.png', dpi=150, bbox_inches='tight')
plt.show()

# Plot 5: Risk by Education Level 
edu_risk = df.groupby('Required Education')['Automation Risk (%)'].mean().sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 5))
ax.barh(range(len(edu_risk)), edu_risk.values,
        color=plt.cm.Spectral(edu_risk.values / 100))
ax.set_yticks(range(len(edu_risk)))
ax.set_yticklabels(edu_risk.index)
ax.set_xlabel('Average Automation Risk (%)')
ax.set_title('Average Automation Risk by Education Level')
ax.grid(axis='x', alpha=0.3)

for i, v in enumerate(edu_risk.values):
    ax.text(v + 0.5, i, f'{v:.1f}%', va='center', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/05_risk_by_education.png', dpi=150, bbox_inches='tight')
plt.show()

#  Plot 6: Risk by Job Status 
status_risk = df.groupby('Job Status')['Automation Risk (%)'].mean()

# Green for growing jobs, red for declining jobs
bar_colors = ['#2ecc71' if s == 'Increasing' else '#e74c3c' for s in status_risk.index]

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(status_risk.index, status_risk.values, color=bar_colors, edgecolor='black')
ax.set_ylabel('Average Automation Risk (%)')
ax.set_title('Average Automation Risk by Job Status')
ax.grid(axis='y', alpha=0.3)

for i, v in enumerate(status_risk.values):
    ax.text(i, v + 0.5, f'{v:.1f}%', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/06_risk_by_job_status.png', dpi=150, bbox_inches='tight')
plt.show()

#  Plots 7-9: Scatter plots (numerical vs Automation Risk) 
# We use a helper function to avoid repeating the same code 3 times
def scatter_vs_risk(x_col, color_map, bg_color, save_name):
    """Plot x_col vs Automation Risk as a scatter plot with correlation label."""
    fig, ax = plt.subplots(figsize=(10, 5))

    sc = ax.scatter(
        df[x_col],
        df['Automation Risk (%)'],
        c=df['Automation Risk (%)'],   # colour each dot by its risk value
        cmap=color_map,
        alpha=0.6, s=25, edgecolors='black', linewidth=0.3
    )
    plt.colorbar(sc, ax=ax, label='Automation Risk (%)')

    # Pearson correlation: +1 = strong positive, -1 = strong negative, 0 = no relation
    corr = df[x_col].corr(df['Automation Risk (%)'])
    ax.text(0.05, 0.95, f'Correlation: {corr:.3f}',
            transform=ax.transAxes, fontsize=11, verticalalignment='top',
            bbox=dict(boxstyle='round', facecolor=bg_color, alpha=0.8))

    ax.set_xlabel(x_col)
    ax.set_ylabel('Automation Risk (%)')
    ax.set_title(f'{x_col} vs Automation Risk')
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.savefig(f'{OUTPUT_DIR}/{save_name}', dpi=150, bbox_inches='tight')
    plt.show()


scatter_vs_risk('Median Salary (USD)',          'RdYlGn_r', 'wheat',        '07_salary_vs_risk.png')
scatter_vs_risk('Experience Required (Years)',  'coolwarm',  'lightblue',   '08_experience_vs_risk.png')
scatter_vs_risk('Remote Work Ratio (%)',         'viridis',  'lightyellow', '09_remote_work_vs_risk.png')

print("Bivariate analysis done!")


print("\n" + "=" * 65)
print("MULTIVARIATE ANALYSIS")
print("=" * 65)

# ── Plot 10: Correlation heatmap ───────────────────────────────────────────
# Shows how every pair of numerical columns relate to each other
# +1 = move together, -1 = move opposite, 0 = no relationship
corr_matrix = df[numerical_cols].corr()

fig, ax = plt.subplots(figsize=(10, 8))
sns.heatmap(
    corr_matrix,
    annot=True,       # show numbers inside each cell
    fmt='.2f',        # 2 decimal places
    cmap='coolwarm',  # blue=negative, red=positive
    center=0,
    square=True,
    linewidths=0.5,
    vmin=-1, vmax=1,
    ax=ax
)
ax.set_title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/10_correlation_matrix.png', dpi=150, bbox_inches='tight')
plt.show()

# Print which features correlate most with Automation Risk
print("\nCorrelations with Automation Risk (strongest first):")
print("-" * 55)
target_corr = (corr_matrix['Automation Risk (%)']
               .drop('Automation Risk (%)')
               .sort_values(key=abs, ascending=False))
for feature, val in target_corr.items():
    print(f"  {feature:<35s}: {val:+.3f}")

# ── Plot 11: Pairplot ──────────────────────────────────────────────────────
# Draws a scatter plot for every combination of the selected columns.
# Diagonal shows each column's own distribution.
key_features = [
    'Automation Risk (%)',
    'Median Salary (USD)',
    'Experience Required (Years)',
    'Remote Work Ratio (%)'
]

# Sample 1000 rows so the plot renders quickly
sample_df = df[key_features].sample(n=min(1000, len(df)), random_state=42)

print("\nGenerating pairplot (may take a moment)...")
pairplot = sns.pairplot(sample_df, diag_kind='kde', plot_kws={'alpha': 0.5}, corner=True)
pairplot.fig.suptitle('Pairplot of Key Features (1000-row sample)', y=1.02)
plt.savefig(f'{OUTPUT_DIR}/11_pairplot.png', dpi=150, bbox_inches='tight')
plt.show()

# ── Plot 12: Job Growth vs Risk ────────────────────────────────────────────
# Positive growth = more jobs expected by 2030, negative = shrinking field
df['Job_Growth_%'] = (
    (df['Projected Openings (2030)'] - df['Job Openings (2024)'])
    / df['Job Openings (2024)']
    * 100
)

fig, ax = plt.subplots(figsize=(12, 6))
sc = ax.scatter(
    df['Job_Growth_%'], df['Automation Risk (%)'],
    c=df['Automation Risk (%)'], cmap='RdYlGn_r',
    alpha=0.6, s=40, edgecolors='black', linewidth=0.3
)
plt.colorbar(sc, ax=ax, label='Automation Risk (%)')

ax.axhline(y=50, color='red',  linestyle='--', linewidth=1.5, label='50% risk threshold')
ax.axvline(x=0,  color='blue', linestyle='--', linewidth=1.5, label='No growth / decline')
ax.set_xlabel('Job Growth Rate (%) 2024–2030')
ax.set_ylabel('Automation Risk (%)')
ax.set_title('Job Growth vs Automation Risk')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/12_job_growth_vs_risk.png', dpi=150, bbox_inches='tight')
plt.show()

# ── Plot 13: Industry summary – 4 metrics side by side ────────────────────
industry_summary = df.groupby('Industry').agg(
    avg_risk       = ('Automation Risk (%)',   'mean'),
    avg_salary     = ('Median Salary (USD)',   'mean'),
    avg_remote     = ('Remote Work Ratio (%)', 'mean'),
    avg_job_growth = ('Job_Growth_%',          'mean')
).round(1)

print("\nIndustry Summary:")
print(industry_summary)

fig, axes = plt.subplots(2, 2, figsize=(16, 12))

industry_summary['avg_risk'].sort_values().plot(kind='barh', ax=axes[0, 0], color='coral')
axes[0, 0].set_xlabel('Average Automation Risk (%)')
axes[0, 0].set_title('Automation Risk by Industry')
axes[0, 0].grid(axis='x', alpha=0.3)

industry_summary['avg_salary'].sort_values().plot(kind='barh', ax=axes[0, 1], color='lightblue')
axes[0, 1].set_xlabel('Average Salary (USD)')
axes[0, 1].set_title('Salary by Industry')
axes[0, 1].grid(axis='x', alpha=0.3)

industry_summary['avg_remote'].sort_values().plot(kind='barh', ax=axes[1, 0], color='lightgreen')
axes[1, 0].set_xlabel('Average Remote Work Ratio (%)')
axes[1, 0].set_title('Remote Work Ratio by Industry')
axes[1, 0].grid(axis='x', alpha=0.3)

industry_summary['avg_job_growth'].sort_values().plot(kind='barh', ax=axes[1, 1], color='plum')
axes[1, 1].set_xlabel('Average Job Growth (%)')
axes[1, 1].set_title('Job Growth by Industry (2024–2030)')
axes[1, 1].axvline(x=0, color='red', linestyle='--', linewidth=1.5)  # zero-growth line
axes[1, 1].grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/13_industry_comprehensive.png', dpi=150, bbox_inches='tight')
plt.show()

print("Multivariate analysis done!")


# =============================================================================
# STEP 6 - KEY INSIGHTS
# =============================================================================

print("\n" + "=" * 65)
print("KEY INSIGHTS")
print("=" * 65)

# Calculate each insight
avg_risk          = df['Automation Risk (%)'].mean()
industry_risk     = df.groupby('Industry')['Automation Risk (%)'].mean()
highest_industry  = industry_risk.idxmax()
edu_risk          = df.groupby('Required Education')['Automation Risk (%)'].mean().sort_values()
status_risk       = df.groupby('Job Status')['Automation Risk (%)'].mean()
salary_corr       = df['Median Salary (USD)'].corr(df['Automation Risk (%)'])
remote_corr       = df['Remote Work Ratio (%)'].corr(df['Automation Risk (%)'])
high_risk_count   = (df['Automation Risk (%)'] > 80).sum()
low_risk_count    = (df['Automation Risk (%)'] < 30).sum()
total             = len(df)

# Print all insights
print(f"1.  Average automation risk across all jobs  : {avg_risk:.1f}%")
print(f"2.  Industry with highest risk               : {highest_industry} ({industry_risk.max():.1f}%)")
print(f"3.  Education with LOWEST risk               : {edu_risk.index[0]} ({edu_risk.values[0]:.1f}%)")
print(f"4.  Education with HIGHEST risk              : {edu_risk.index[-1]} ({edu_risk.values[-1]:.1f}%)")
print(f"5.  Avg risk for INCREASING jobs             : {status_risk.get('Increasing', float('nan')):.1f}%")
print(f"6.  Avg risk for DECREASING jobs             : {status_risk.get('Decreasing', float('nan')):.1f}%")
print(f"7.  Salary  <-> Automation Risk correlation  : {salary_corr:+.3f}")
print(f"8.  Jobs with >80% automation risk           : {high_risk_count} ({high_risk_count/total*100:.1f}%)")
print(f"9.  Jobs with <30% automation risk           : {low_risk_count} ({low_risk_count/total*100:.1f}%)")
print(f"10. Remote Work <-> Automation Risk corr     : {remote_corr:+.3f}")
print("=" * 65)

# Save insights to a text file
with open(f'{OUTPUT_DIR}/insights.txt', 'w') as f:
    f.write("KEY INSIGHTS FROM EDA\n")
    f.write("=" * 65 + "\n")
    f.write(f"Average automation risk            : {avg_risk:.1f}%\n")
    f.write(f"Highest-risk industry              : {highest_industry}\n")
    f.write(f"Salary correlation with risk       : {salary_corr:+.3f}\n")
    f.write(f"Remote work correlation with risk  : {remote_corr:+.3f}\n")

print(f"\nInsights saved to {OUTPUT_DIR}/insights.txt")
print(f"All plots saved in: {OUTPUT_DIR}")
print("\nEDA complete! You can now proceed to Feature Engineering.")
```
